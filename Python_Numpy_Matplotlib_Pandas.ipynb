### Ejercicio 1

Crea un array de NumPy 2D como este: [[1, 2, 3, 4], [5, 6, 7, 8]], y, a continuación:
- a) Consulta el número de dimensiones, la longitud de estas, y el número total de elementos. Además, muestra los valores de la posicion (1,3), y los 3 ultimos elementos de la segunda dimensión. **(0.25 puntos)**
- b) Itera a través de los elementos del array multidimensional para mostrar por pantalla, uno a uno, los elementos que son múltiplos de 3. **(0.25 puntos)**

**(Total 0.5 puntos)** $\color{green}{\text{NM}}$

```python
# Crear el array 2D
array = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

# a) Consultar la información del array
print("Número de dimensiones:", array.ndim)
print("Longitud del array(filas, columnas):", array.shape)
print("Número total de elementos:", array.size)

# Valor en la posición (1, 3)
print("Valor en la posición (1, 3):", array[1, 3])

# Últimos elementos de la segunda dimensión (segunda fila)
print("Tres últimos elementos de la segunda fila:", array[1, -3:])

# b) Mostrar los elementos múltiplos de 3
print("\nElementos múltiplos de 3:")
for fila in array:
    for elemento in fila:
        if elemento % 3 == 0:
            print(elemento)
```

### Ejercicio 3

Crea un array 2D de dimensiones (11, 11) de ceros, y cambia a 1 los elementos necesarios para dibujar los siguientes patrones:

- a) lineas verticales equidistantes **(0.5 puntos)**
- b) una cruz **(0.5 puntos)**
- c) un tablero de ajedrez **(0.5 puntos)**
- d) una x **(0.5 puntos)**

**Nota**. Para visualizar la matriz tienes que buscar qué función de `Matplotlib` te permite visualizar la matriz como una imagen en 2D.

**(Total 2 puntos)** $\color{Orange}{\text{EG}}$

```python
# Importar matplotlib con la abreviatura "plt"
import matplotlib.pyplot as plt
```

```python
# a)---------------------------------------------
# Crear matriz 11x11 de ceros
a = np.zeros((11, 11), dtype=int)

# Poner 1 en columnas equidistantes (cada 2 columnas)
a[:, ::2] = 1  # desde la columna 0 cada 3 posiciones

# Visualizar
plt.imshow(a, cmap='gray')
plt.title("a) Líneas verticales equidistantes")
plt.show()
```

```python
# b)-------------------------------------------
# Crear nueva matriz de ceros
b = np.zeros((11, 11), dtype=int)

# Cruz: fila central y columna central
b[5, :] = 1       # fila central
b[:, 5] = 1       # columna central

# Visualizar
plt.imshow(b, cmap='gray')
plt.title("b) Una cruz")
plt.show()
```

```python
# c)-------------------------------------------
# np.indices genera las coordenadas (fila, columna) de cada posición
# Al sumar los índices y hacer % 2, se alternan ceros y unos
c = np.indices((11, 11)).sum(axis=0) % 2

# Visualizar
plt.imshow(c, cmap='gray')
plt.title("c) Tablero de ajedrez")
plt.show()
```

```python
# d)-------------------------------------------------
# Crear matriz de ceros
d = np.zeros((11, 11), dtype=int)

# Diagonal principal
np.fill_diagonal(d, 1)

# Diagonal secundaria (volteo columnas)
np.fill_diagonal(np.fliplr(d), 1)

# Visualizar
plt.imshow(d, cmap='gray')
plt.title("d) Una X")
plt.show()
```

### Ejercicio 4

Imagina que tienes tres cuerpos en un plano bidimensional.  Las fuerzas que actuan sobre cada cuerpo se pueden representar en una matriz.  Cada fila de la matriz representa un cuerpo y cada columna representa una fuerza en una dirección concreta ( por ejemplo, fuerza en la dirección X y fuerza en la dirección Y) en Newtons.

Las fuerzas que actuan sobre los cuerpos son las seguientes:
- Cuerpo A: Fuerza X = 10 y Fuerza Y = 5.
- Cuerpo B: Fuerza X = 4 y Fuerza Y = 12.
- Cuerpo C: Fuerza X = -6 y Fuerza Y = -8.

Responde a las seguientes preguntas:
- a) Crea la matriz de fuerzas y muéstrala por pantalla. **(0.25 puntos)**
- b) Calcula la fuerza resultante sobre cada cuerpo. La fuerza resultante se puede calcular como la suma vectorial de las fuerzas.  Para hacerlo, utiliza una función de `numpy`. **(0.50 puntos)**
- c) Calcula la aceleración de cada cuerpo mediante la segunda ley de Newton, `F=m*a`, considerando que la masa de cada cuerpo es de 5 kg. **(0.25 puntos)**

**Nota**.  Los resultados deben presentarse con sólo dos decimales.

**(Total 1 punto)** $\color{Orange}{\text{EG}}$

```python
# a) Crear la matriz de fuerzas (cada fila es un cuerpo, columnas = Fx y Fy)
fuerzas = np.array([
    [10, 5],   # Cuerpo A
    [4, 12],   # Cuerpo B
    [-6, -8]   # Cuerpo C
])
print("a) Matriz de fuerzas (en N):")
print(fuerzas)

# b) Calcular la fuerza resultante por cuerpo (módulo del vector)
# Fórmula: ||F|| = sqrt(Fx² + Fy²)
modulos_fuerza = np.linalg.norm(fuerzas, axis=1)
modulos_fuerza = np.round(modulos_fuerza, 2)

print("\nb) Fuerza resultante por cuerpo (N):")
print(modulos_fuerza)

# c) Calcular la aceleración usando la segunda ley de Newton: F = m * a
# a = F / m
masa = 5  # kg
aceleraciones = modulos_fuerza / masa
aceleraciones = np.round(aceleraciones, 2)

print("\nc) Aceleración por cuerpo (m/s²):")
print(aceleraciones)
```

**(Total 4 puntos)**  

```python
#1. Leer el archivo "sales"
ruta = "/content/drive/MyDrive/Colab Notebooks/Activity_3/data/sales.csv"
df = pd.read_csv(ruta)

# Mostrar primeras filas
print("Archivo 'sales.csv' cargado correctamente.")
print(df.head())
```

```python
# a.1) Mostrar 8 primeras y 5 últimas filas,
# y dimensiones del DataFrame
print("a.1) Primeras 8 filas:")
print(df.head(8))

print("\na.1) Últimas 5 filas:")
print(df.tail(5))

print(f"\nEl DataFrame tiene {df.shape[0]} filas y {df.shape[1]} columnas.")

# a.2) Mostrar qué  tipo de variable es cada columna
#  y convertir columna 'Date' a datetime
print("\na.2) Tipos de variables antes de conversión:")
print(df.dtypes)

df["Date"] = pd.to_datetime(df["Date"])  # conversión de fecha

print("\nTipos de variables después de conversión:")
print(df.dtypes)

# a.3) Estadísticas descriptivas de variables numéricas
print("\na.3) Estadísticas descriptivas:")
print(df.describe(percentiles=[0.25, 0.5, 0.75]))

# a.4) Crear columna 'Total_Sales' = Quantity * Price
df["Total_Sales"] = df["Quantity"] * df["Price"]

print("\na.4) Columna 'Total_Sales' añadida correctamente:")
print(df[["Quantity", "Price", "Total_Sales"]].head())
```

```python
# Calcula las ventas totales por producto
#y visualiza el resultado mediante un gráfico de barras.

#1. Agrupar por producto y sumar ventas
ventas_por_producto = df.groupby("Product")["Total_Sales"].sum()

# Crear gráfico de barras con colores diferentes
plt.figure(figsize=(10, 6))
plt.bar(
    ventas_por_producto.index,                # productos en el eje X
    ventas_por_producto.values,               # valores de ventas en eje Y
    color=plt.cm.tab10.colors[:len(ventas_por_producto)]  # colores distintos
)

# Título y etiquetas
plt.title("Ventas Totales por producto")
plt.xlabel("Producto")
plt.ylabel("Ventas Totales")
plt.xticks(rotation=45)  # rotar etiquetas si son muchas
plt.tight_layout()  # para que no se solapen los textos
plt.show()
```

```python
# Cálculo de las ventas totales agrupadas por meses
# y visualizar la evolución temporal mediante un gráfico de línea.

# Crear una nueva columna con el mes (formato AAAA-MM)
df["Mes"] = df["Date"].dt.to_period("M")

# Agrupar por mes y calcular ventas totales
ventas_mensuales = df.groupby("Mes")["Total_Sales"].sum()

# Convertir PeriodIndex a string para el eje X
ventas_mensuales.index = ventas_mensuales.index.astype(str)

# Crear gráfico de línea verde
plt.figure(figsize=(10, 6))
plt.plot(ventas_mensuales.index, ventas_mensuales.values,
         marker='o', linestyle='-', color='green')

# Añadir título y etiquetas
plt.title("Ventas Totales a lo largo del tiempo")
plt.xlabel("Mes")
plt.ylabel("Ventas Totales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Calcula la distribución de ventas por diferentes regiones
# y visualiza el resultado con un gráfico circular.

# Agrupar ventas totales por región
ventas_por_region = df.groupby("Region")["Total_Sales"].sum()

# Crear gráfico circular (pie chart)
plt.figure(figsize=(8, 8))
plt.pie(
    ventas_por_region,
    labels=ventas_por_region.index,
    autopct='%1.1f%%',         # mostrar porcentajes
    startangle=90              # empieza desde arriba
)

# Añadir título
plt.title("Distribución de ventas por región")
plt.tight_layout()
plt.show()
```

En el siguiente ejercicio utilizaremos la librería `SciPy` para verificar si un conjunto de datos sigue una distribución normal.
   - a) Crea un conjunto de datos aleatorios con 50 valores que sigan una distribución normal (media de 100 y desviación estándar de 15). **(0.50 puntos)** $\color{green}{\text{NM}}$
   - b) Realiza un test de normalidad utilizando la función [shapiro](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.shapiro.html) para determinar si el conjunto de datos sigue una distribución normal. Muestra por pantalla el p-value y razona el resultado obtenido. **(1 punto)** $\color{orange}{\text{EG}}$
   - c) Crea un histograma del conjunto de datos para visualizar su distribución. **(0.50 puntos)** $\color{green}{\text{NM}}$  
   - d) Utiliza una función de la librería `seaborn` para visualizar la función de densidad de probabilidad correspondiente a la distribución de los datos creados. **Opcional**. $\color{red}{\text{EI}}$  
   


**(Total 2 puntos)**

```python
# Importar librerias scipy y shapiro
from scipy.stats import shapiro
import seaborn as sns
```

```python
# a) Crear un conjunto de 50 valores con distribución normal (media=100, std=15)
np.random.seed(176)
datos = np.random.normal(loc=100, scale=15, size=50)

print("a) Muestra de los primeros diez valoresa:")
print(np.round(datos[:10], 2))
```

```python
# b) Test de normalidad de Shapiro-Wilk
stat, p_value = shapiro(datos)

print("\nb) Test de Shapiro-Wilk:")
print(f"Estadístico: {stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretación
if p_value > 0.05:
    print("Conclusión: No se rechaza la hipótesis nula. Los datos siguen una distribución normal.")
else:
    print("Conclusión: Se rechaza la hipótesis nula. Los datos NO siguen una distribución normal.")
```

```python
# c) Histograma de los datos
plt.figure(figsize=(8, 5))
plt.hist(datos, bins=10, color="skyblue", edgecolor="black")
plt.title("c) Histograma de los datos generados")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# d) Visualización de la función de densidad con seaborn
plt.figure(figsize=(8, 5))
sns.kdeplot(datos, fill=True, color="purple")
plt.title("d) Función de densidad de probabilidad (Seaborn)")
plt.xlabel("Valor")
plt.ylabel("Densidad")
plt.grid(True)
plt.tight_layout()
plt.show()
```
